# if else elif

```python
marks=(int(input("How much marks you get in Data science")))
if marks>=90:
    print("Grade A")
elif marks>=80<=90:
    print("Grade B")
elif marks>=70<80:
    print("Grade C")
elif marks>=60<70:
    print("Grade D")
else:
    print("Fail")
```
